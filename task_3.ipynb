```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
population = np.random.exponential(scale=2.0, size=50000)

n1, n2 = 10, 50
samples_count = 1000

means_n1 = [np.mean(np.random.choice(population, n1)) for _ in range(samples_count)]
means_n2 = [np.mean(np.random.choice(population, n2)) for _ in range(samples_count)]

print(f"n={n1} -> Середнє: {np.mean(means_n1):.4f}, Std: {np.std(means_n1):.4f}")
print(f"n={n2} -> Середнє: {np.mean(means_n2):.4f}, Std: {np.std(means_n2):.4f}")
```

```python
fig, axs = plt.subplots(1, 2, figsize=(12, 5))

axs[0].hist(means_n1, bins=30, color='blue', alpha=0.7, edgecolor='black')
axs[0].set_title(f'Розподіл вибіркових середніх (n={n1})')
axs[0].set_xlabel('Вибіркове середнє')
axs[0].set_ylabel('Частота')

axs[1].hist(means_n2, bins=30, color='green', alpha=0.7, edgecolor='black')
axs[1].set_title(f'Розподіл вибіркових середніх (n={n2})')
axs[1].set_xlabel('Вибіркове середнє')

plt.tight_layout()
plt.show()
```
