```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
dates_ts = pd.date_range(start='2023-01-01', periods=90)
sales = np.random.randint(50, 300, 90)
df_ts = pd.DataFrame({'date': dates_ts, 'sales': sales})

window_size = 7
df_ts['rolling_mean'] = df_ts['sales'].rolling(window=window_size).mean()
df_ts['rolling_std'] = df_ts['sales'].rolling(window=window_size).std()

print("Перші 10 рядків таблиці:")
print(df_ts.head(10))
```

```python
fig, axs = plt.subplots(2, 1, figsize=(10, 8), sharex=True)

axs[0].plot(df_ts['date'], df_ts['sales'], color='lightgray', label='Вихідні продажі', marker='.')
axs[0].plot(df_ts['date'], df_ts['rolling_mean'], color='red', label=f'Ковзне середнє ({window_size} днів)', linewidth=2)
axs[0].set_title('Динаміка продажів та ковзне середнє')
axs[0].set_ylabel('Продажі')
axs[0].legend()
axs[0].grid(True, linestyle='--', alpha=0.6)

axs[1].plot(df_ts['date'], df_ts['rolling_std'], color='orange', label=f'Ковзне StdDev ({window_size} днів)')
axs[1].set_title('Ковзне стандартне відхилення')
axs[1].set_xlabel('Дата')
axs[1].set_ylabel('StdDev')
axs[1].legend()
axs[1].grid(True, linestyle='--', alpha=0.6)
axs[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```
